# Global Earthquake–Tsunami Risk Assessment
This notebook answers the project exploratory questions using the provided dataset.

```
import pandas as pd

df = pd.read_csv('/mnt/data/earthquake_data_tsunami.csv')
df.head()
```

## Question 1: How does magnitude affect tsunami likelihood?

```
df.groupby('tsunami')['magnitude'].describe()
```

## Question 2: Geographic distribution of tsunami events

```
import matplotlib.pyplot as plt

plt.scatter(df['longitude'], df['latitude'], s=5)
plt.xlabel('Longitude'); plt.ylabel('Latitude'); plt.title('Earthquake Locations');
plt.show()
```

## Question 3: Depth category tsunami frequency

```
def depth_category(d):
    if d < 50: return 'shallow'
    if d < 300: return 'intermediate'
    return 'deep'

df['depth_cat'] = df['depth'].apply(depth_category)
df.groupby('depth_cat')['tsunami'].mean()
```

## Question 4: Time-based patterns in tsunami frequency

```
annual = df.groupby('Year')['tsunami'].sum()
plt.plot(annual.index, annual.values)
plt.xlabel('Year'); plt.ylabel('Tsunami Count'); plt.title('Annual Tsunami Events');
plt.show()
```
